```python
# Test reading file
# companies = ' '.join(open("companies.txt", encoding="utf-8").read().splitlines())
# companies

def get_with_space(c: str): # change the name of this function
    return "<S>" if c == " " else c
```

```python
# CORPORA & LOOK UP TABLE

# Loading corpora
corpora = ''.join(open("any.txt", encoding="utf-8").read().splitlines())
# print(corpora)

# Char based model
vocabulary = sorted(set(corpora))
vocabulary[0] = "<S>"
# print(vocabulary)

look_up_ctoi = {c: i+1 for i,c in enumerate(vocabulary)}
look_up_ctoi["<.>"] = 0 
print(look_up_ctoi)

look_up_itoc = {i: c for c, i in look_up_ctoi.items()}
print(look_up_itoc)

# Different lenght can make things a bit confuse (?)
# print(len(vocabulary)) # 54
# print(len(look_up_ctoi)) # 55
# print(len(look_up_itoc))

data = []
labels = []
# corpora[0]

for i, xy in enumerate(zip(corpora[:-1], corpora[2:])):
    xs = [look_up_ctoi[get_with_space(xy[0])], look_up_ctoi[get_with_space(corpora[i+1])]]
    data.append(xs)
    ys = look_up_ctoi[get_with_space(xy[1])]
    labels.append(ys)

# print(data)
# print(labels)

# print(len(data))
# len(labels)

xs = torch.tensor(data)
ys = torch.tensor(labels)

print(xs[:20])
ys[:10]

xs.shape


```
