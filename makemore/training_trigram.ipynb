```python
# Utils


class TextCorpus:

    _corpus: list
    _content: str 
    _trigrams: list[tuple[int, int, int]]

    stoi: dict[str, int] 

    def __init__(self, content: str) -> None:
        if not len(content) >= 3:
            raise ValueError()
        
        self._content = content
        self._trigrams = []
        
        self._build_corpus()

        
    def _build_corpus(self):
        self._corpus = sorted(list(set(self._content)))
        self.stoi = { t: i for i, t in enumerate(self._corpus)}

    @property
    def corpus_size(self):
        return len(self._corpus)
    
    @property
    def corpus(self):
        return self._corpus.copy()
    
    def add_trigram(self, trigram: tuple[int, int, int]):
        self._trigrams.append(trigram)

    def replace_empty_space(self, c: str): 
        return "<SPACE>" if c == " " else c
    
    def __str__(self) -> str:
        return ', '.join(self._corpus)
```
