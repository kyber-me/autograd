```python
# Imports

import torch
import numpy as np

```

```python
# Utils
def get_with_special_char(c: str): 
    return "<SPACE>" if c == " " else c
```

```python
# Loading corpora
corpora = ''.join(open("any.txt", encoding="utf-8").read().splitlines())
print("corpora: ", corpora)

# Char based model
vocabulary = sorted(set(corpora))

# When it sorts the chars empty space becomes the first one
vocabulary[0] = "<SPACE>"
vocabulary.extend(["<START>", "<END>"])
print("vocabulary: ", vocabulary)

look_up_ctoi = {c: i for i,c in enumerate(vocabulary)}
print(look_up_ctoi)

look_up_itoc = {i: c for c, i in look_up_ctoi.items()}
# print(look_up_itoc)


data = []
labels = []
raw_data = []
raw_labels = []

# size = len(corpora)
for i, xy in enumerate(zip(corpora[:-1], corpora[1:], corpora[2:])):
    if i == 0:
        raw_input = ["<START>", xy[0]]
        raw_output = xy[1]
        raw_data.append(raw_input)
        raw_labels.append(raw_output)

        xs = [look_up_ctoi[key] for key in map(get_with_special_char, raw_input)]
        # xs = [look_up_ctoi[get_with_special_char("<START>")], look_up_ctoi[get_with_special_char(xy[0])]]
        ys = look_up_ctoi[get_with_special_char(raw_output)]
        data.append(xs)    
        labels.append(ys)
    elif i == 1992:
        raw_input = [xy[1], xy[2]]
        raw_output = "<END>"
        raw_data.append(raw_input)
        raw_labels.append(raw_output)

        xs = [look_up_ctoi[key] for key in map(get_with_special_char, raw_input)]
        # xs = [look_up_ctoi[get_with_special_char(xy[1])], look_up_ctoi[get_with_special_char(xy[2])]]
        ys = look_up_ctoi[get_with_special_char(raw_output)]
        data.append(xs)    
        labels.append(ys)
    
    xs = [look_up_ctoi[get_with_special_char(xy[0])], look_up_ctoi[get_with_special_char(xy[1])]]
    data.append(xs)
    ys = look_up_ctoi[get_with_special_char(xy[2])]
    labels.append(ys)

print("Data: ", data, " => ", len(data))
print("Labels: ", labels, " => ", len(labels))

# print(len(data))
# len(labels)

# xs = torch.tensor(data)
# ys = torch.tensor(labels)

# print(xs[:20])
# ys[:10]

# xs.shape


```
