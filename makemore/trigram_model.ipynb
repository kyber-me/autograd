```python
# Abstractions
class TextCorpus:

    _vocabulary: list
    _content: str 
    _trigrams: list[tuple[int, int, int]]

    stoi: dict[str, int] 

    def __init__(self, content: str) -> None:
        if not len(content) >= 3:
            raise ValueError()
        
        self._content = content
        self._trigrams = []
        
        self._build_vocabulary()

        
    def _build_vocabulary(self):
        self._vocabulary = sorted(list(set(self._content)))
        self.stoi = { t: i for i, t in enumerate(self._vocabulary)}

    @property
    def corpus_size(self):
        return len(self._vocabulary)
    
    @property
    def corpus(self):
        return self._vocabulary.copy()
        
    def __str__(self) -> str:
        return ', '.join(self._vocabulary)
    
    def __len__(self):
        return len(self._vocabulary)

    # def replace_empty_space(self, c: str): 
    #     return "<SPACE>" if c == " " else c
```

```python

corpus: TextCorpus | None  
d = str()
with open("different versions of word vectors.txt", 'r', encoding='utf-8') as f:
    d = f.read()
    corpus = TextCorpus(d)
    

# Need to add the `special characters` for Start and End
if len(d) <= 1:
    raise Exception("d is empty.")


xs = torch.zeros((2014, 110))
ys = torch.zeros((2014, 55))
for i, (c1, c2, c3) in enumerate(zip(d[:-1], d[1:], d[2:])):
    c1_pos = corpus.stoi[c1]
    c2_pos = corpus.stoi[c2]
    c3_pos = corpus.stoi[c3]
    xs[i][c1_pos] = 1.
    xs[i][c2_pos+55] = 1.
    ys[i][c3_pos] = 1.





    

```

```python
# from torch._tensor import Tensor
from torch.nn.functional import softmax

g = torch.Generator().manual_seed(42)
W = torch.randn((110, 55), generator=g)
# for i, (x, y) in enumerate(zip(xs, ys)):    
#     target = y 
#     logits = x @ W
print("Weight Matrix")
print(W, end="\n\n\n")

x = xs[0]
print("Input")
print(x, end="\n\n")

y = ys[0]
print("Target")
print(y, end="\n--------------------\n\n\n")

logits = x @ W
print("LOGITS")
print(logits, end="\n\n")

print("PROBS")
prob_dist = softmax(logits, -1)
print(prob_dist, end="\n\n")

filtered_probs = y * prob_dist
print("Filtered probs")
print(filtered_probs, end="\n\n")


# mle = torch.sum(torch.tensor([1., 2.]))
# logprobs = torch.log(y * prob_dist) -> Need copy for those input tensors? And use `out` from the function .log?

logprob = -torch.log(torch.sum(y*prob_dist)) # logprob and MLE (Maximum Estimation Likelihood are the same thing here)
logprob




# print("Result")
# result 

# maximum_likelihook_estimation

```
